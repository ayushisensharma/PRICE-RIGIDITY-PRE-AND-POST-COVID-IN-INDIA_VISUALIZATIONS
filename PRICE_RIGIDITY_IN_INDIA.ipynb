```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as tic
```

```python
data = pd.read_excel("Macro_data.xlsx")
data["Year"] = data["Year"].astype(int)
gdp = data['GDP growth (annual %)']
inflation = data['Inflation, consumer prices (annual %)']
unemployment = data['Unemployment, total (% of total labor force) (modeled ILO estimate)']
Year = data['Year']
change_unemployment = data['Absolute_change_in unemployment']
change_inflation = data['Absolute_change_in_inflation']
adjustment_ratio = data['Adjustment Ratio']
data = data.dropna()
print(data.head())
```

```python
plt.figure()

plt.plot(Year, gdp, label="GDP Growth")
plt.plot(Year, inflation, label="Inflation")
plt.plot(Year, unemployment, label="Unemployment")

plt.axvline(x=2020, linestyle='--',color ="black")

plt.gca().xaxis.set_major_locator(ticker.MaxNLocator(integer=True))

plt.legend()
plt.xlabel("Year")
plt.ylabel("Value (%)")
plt.title("Macroeconomic Indicators")

plt.show()
```

```python
plt.figure()

plt.plot(Year, change_inflation, label="Absolute Change in Inflation Rate")
plt.plot(Year, change_unemployment, label="Absolute Change in Unemployment Rate")


plt.axvline(x=2020, linestyle='--',color ="black")

plt.gca().xaxis.set_major_locator(ticker.MaxNLocator(integer=True))

plt.legend(bbox_to_anchor=(1.05, 1),loc="lower left")
plt.xlabel("Year")
plt.ylabel("Value (pp)")
plt.title("Absolute Change")

plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

plt.figure()

plt.plot(Year,adjustment_ratio, label="Adjustment Ratio")
plt.axvline(x=2020, linestyle='--',color ="black")

plt.gca().xaxis.set_major_locator(ticker.MaxNLocator(integer=True))

plt.legend()
plt.xlabel("Year")
plt.ylabel("Ratio")
plt.title("Adjustment Ratio")

plt.show()
```
